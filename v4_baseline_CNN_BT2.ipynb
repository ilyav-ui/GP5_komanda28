```python
class BaselineCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(16, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 14 * 14, 256),
            nn.ReLU(),
            nn.Linear(256, num_classes)
        )
    def forward(self, x):
        x = self.features(x)
        x = self.classifier(x)
        return x
```

```python
baseline_config = {
    "model_name": "baseline_cnn",
    "model_title": "Baseline CNN",
    "architecture": "simple CNN",
    "image_size": 224,
    "batch_size": batch_size,
    "num_epochs": 10,
    "learning_rate": 0.001,
    "optimizer": "Adam",
    "loss_function": "CrossEntropyLoss",
    "num_classes": num_classes,
    "seed": 42
}

baseline_path = MODELS_DIR / "baseline_cnn_best.pth"
```

```python
baseline_model = BaselineCNN(num_classes).to(device)

baseline_result = train_model(
    baseline_model,
    train_loader,
    val_loader,
    test_loader,
    train_dataset.classes,
    baseline_config,
    baseline_path
)

print("Best val accuracy:", round(baseline_result["best_val_accuracy"], 4))
print("Test accuracy:", round(baseline_result["test_accuracy"], 4))
print("Test macro F1:", round(baseline_result["test_macro_f1"], 4))
print("Test weighted F1:", round(baseline_result["test_weighted_f1"], 4))
```
